Atividade 4

```python
number = int(input("Me diga a quantidade de termos: "))
Fibonacci = [1,1]
for i in range(2, number):
  Fibonacci.append(Fibonacci[i-1] + Fibonacci [i-2])
print(Fibonacci[:number])
```

Atividade 5

```python
while True:
  name = input("Digite seu nome: ")
  if len(name) > 3:
    break
while True:
  age = float(input("Digite sua idade atual: "))
  if 0 <= age <= 150:
    break
while True:
  salary = float(input("Digite seu salário: "))
  if salary > 0:
    break
while True:
  gender = input("Digite seu genero, utilizando F para feminino e M para masculino: ")
  if gender == 'F' or gender == 'M':
    break
while True:
  marital_status = input("Digite seu estado civil, usando s, c, v, d: ")
  if marital_status in ['S', 'C', 'V', 'D']:
    break
print(f"Os dados validados são: Nome = {name}, idade = {age}, salário = {salary:.2f}$, sexo = {gender} e estado civil = {marital_status}.")
```

Atividade 7

```python
number = int(input("Digite um número e eu direi o fatorial dele."))
if number < 0:
  print("Não tem fatorial de números negativos.")
else:
  fatorial = 1
  for i in range(1, number + 1):
    fatorial = fatorial * i
  print(f"O  fatorial desse número será {fatorial}.")
```

Atividade 8

```python
L = [5, 7, 2, 9, 4, 1, 3]
print("O tamanho da lista é: ", len(L))
biggerN = L[0]
for i in L:
  if i > biggerN:
    biggerN = i
smallerN = L[0]
for i in L:
  if i < smallerN:
    smallerN = i
print(f"O maior valor é {biggerN}, o menor valor é {smallerN} e a soma dos valores é {sum(L)}. Em ordem crescente os valores seriam: {sorted(L)}. E em ordem decrescente seriam {sorted(L, reverse=True)}.")
```
